## 3.1 A/B Testing
**Teori (tanpa kode):** A/B test adalah eksperimen dengan **dua kelompok (treatment)** untuk menentukan mana yang lebih baik. Contohnya dua versi halaman web, dua harga, atau dua judul iklan.
- **Treatment**: perlakuan yang diuji (misalnya versi halaman B).
- **Control group**: kelompok yang tidak mendapat perlakuan baru (versi standar A).
- **Randomization**: subjek dibagi ke kelompok **secara acak**. Dengan begitu, perbedaan hasil hanya bisa disebabkan oleh treatment atau oleh kebetulan.
- **Blinding**: subjek (*single-blind*) atau bahkan peneliti (*double-blind*) tidak tahu siapa mendapat treatment yang mana.
- **Test statistic**: metrik yang dibandingkan (misalnya rata-rata waktu sesi atau *conversion rate*). Metrik ini harus ditentukan **sebelum** eksperimen dijalankan.

## 3.2 Hypothesis Tests
**Teori (tanpa kode):** Manusia cenderung meremehkan peran kebetulan dan melihat pola di data acak. Uji hipotesis melindungi kita dari kesalahan ini.
- **Null hypothesis ($H_0$)**: tidak ada perbedaan. Semua perbedaan yang terlihat hanyalah kebetulan.
- **Alternative hypothesis ($H_A$)**: kebalikan dari $H_0$, yaitu ada perbedaan.
- **One-way test**: kita hanya peduli pada satu arah (misalnya "B **lebih baik** dari A").
- **Two-way test**: kita peduli pada perbedaan ke arah mana pun.
- Logikanya seperti *proof by contradiction*: anggap $H_0$ benar, lalu tunjukkan bahwa data yang kita amati **sangat tidak mungkin** muncul jika $H_0$ benar.

## 3.3 Resampling — Permutation Test
**Teori:** **Permutation test** menguji $H_0$ tanpa asumsi distribusi:
1. **Gabungkan** data semua kelompok menjadi satu.
2. **Acak** urutannya, lalu bagi kembali menjadi kelompok dengan ukuran yang sama seperti aslinya.
3. Hitung statistik yang sama (misalnya selisih mean), lalu catat hasilnya.
4. Ulangi langkah 2–3 sebanyak $R$ kali. Hasilnya adalah **distribusi selisih jika $H_0$ benar**.
5. Bandingkan selisih yang **benar-benar teramati** dengan distribusi tersebut. Jika selisih observasi berada di ekor ekstrem, perbedaannya **signifikan secara statistik**.

Perbedaannya dengan bootstrap: permutation test mengacak **tanpa pengembalian** untuk menguji hipotesis, sedangkan bootstrap mengambil sampel **dengan pengembalian** untuk mengukur ketidakpastian estimasi.

**Contoh:** Data `web_page_data.csv` berisi waktu sesi (*session time*) pengunjung di Page A dan Page B. Waktunya dikalikan 100 agar dalam satuan detik. Apakah Page B membuat pengunjung bertahan lebih lama (lebih "sticky")?

````python
session_times = pd.read_csv(WEB_PAGE_DATA_CSV)
session_times.Time = 100 * session_times.Time
````

````python
ax = session_times.boxplot(by='Page', column='Time',
                           figsize=(4, 4))
ax.set_xlabel('')
ax.set_ylabel('Time (in seconds)')
plt.suptitle('')

plt.tight_layout()
plt.show()
````

**Interpretasi:** Boxplot menunjukkan waktu sesi di **Page B cenderung lebih lama** dibanding Page A, tetapi kedua kotak saling tumpang tindih cukup banyak. Apakah perbedaan ini nyata?

````python
mean_a = session_times[session_times.Page == 'Page A'].Time.mean()
mean_b = session_times[session_times.Page == 'Page B'].Time.mean()
print(mean_b - mean_a)
````

**Interpretasi:** Rata-rata waktu sesi Page B lebih lama **35,67 detik** dibanding Page A. Ini adalah *observed difference* yang akan kita uji.

Fungsi permutasi: gabungkan data, ambil acak $n_B$ data sebagai grup B dan sisanya sebagai grup A, lalu hitung selisih mean.

````python
# Permutation test example with stickiness
def perm_fun(x, nA, nB):
    n = nA + nB
    idx_B = set(random.sample(range(n), nB))
    idx_A = set(range(n)) - idx_B
    return x.loc[list(idx_B)].mean() - x.loc[list(idx_A)].mean()
    
nA = session_times[session_times.Page == 'Page A'].shape[0]
nB = session_times[session_times.Page == 'Page B'].shape[0]
print(perm_fun(session_times.Time, nA, nB))
````

````python
random.seed(1)
perm_diffs = [perm_fun(session_times.Time, nA, nB) for _ in range(1000)]

fig, ax = plt.subplots(figsize=(5, 5))
ax.hist(perm_diffs, bins=11, rwidth=0.9)
ax.axvline(x = mean_b - mean_a, color='black', lw=2)
ax.text(50, 190, 'Observed\ndifference', bbox={'facecolor':'white'})
ax.set_xlabel('Session time differences (in seconds)')
ax.set_ylabel('Frequency')

plt.tight_layout()
plt.show()
````

**Interpretasi:** Histogram ini adalah distribusi selisih mean dari 1.000 permutasi acak, yaitu gambaran dunia di mana $H_0$ benar. Garis hitam menandai selisih observasi (35,67 detik), yang posisinya masih berada di dalam sebaran hasil acak, bukan jauh di ekor.

````python
# convert perm_diffs to numpy array to avoid problems with some Python installations
perm_diffs = np.array(perm_diffs)
print(np.mean(perm_diffs > mean_b - mean_a))
````

**Interpretasi:** Sekitar **12,1%** permutasi menghasilkan selisih yang lebih besar dari observasi, sehingga **p-value ≈ 0,12**. Karena 0,12 > 0,05, perbedaan 35,67 detik **tidak signifikan secara statistik**: selisih sebesar itu cukup sering muncul hanya karena kebetulan.

## 3.4 Statistical Significance and P-Values
**Teori:**
- **Statistical significance**: hasil disebut signifikan jika sangat kecil kemungkinannya terjadi karena kebetulan (di bawah $H_0$).
- **p-value**: probabilitas memperoleh hasil **seekstrem atau lebih ekstrem** dari hasil observasi, **dengan asumsi $H_0$ benar**.
- **Alpha ($\alpha$)**: ambang batas yang ditentukan sebelumnya, biasanya 0,05. Jika p-value < α, tolak $H_0$.
- **Type I error**: menyimpulkan ada efek padahal sebenarnya tidak ada (*false positive*).
- **Type II error**: menyimpulkan tidak ada efek padahal sebenarnya ada (*false negative*), biasanya karena sampel terlalu kecil.

**Kesalahpahaman umum:** p-value **bukan** "probabilitas bahwa hasilnya kebetulan" dan **bukan** probabilitas $H_0$ benar. Pada 2016, American Statistical Association (ASA) memperingatkan agar keputusan tidak hanya didasarkan pada p-value < 0,05.

**Contoh:** konversi harga A: 200 dari 23.739 pengunjung (0,8425%), dan harga B: 182 dari 22.588 pengunjung (0,8057%). Selisihnya hanya 0,0368 poin persen. Apakah perbedaan ini signifikan? Kita jawab dengan permutation test.

````python
random.seed(1)
obs_pct_diff = 100 * (200 / 23739 - 182 / 22588)
print(f'Observed difference: {obs_pct_diff:.4f}%')
conversion = [0] * 45945
conversion.extend([1] * 382)
conversion = pd.Series(conversion)

perm_diffs = [100 * perm_fun(conversion, 23739, 22588) 
              for _ in range(1000)]

fig, ax = plt.subplots(figsize=(5, 5))
ax.hist(perm_diffs, bins=11, rwidth=0.9)
ax.axvline(x=obs_pct_diff, color='black', lw=2)
ax.text(0.06, 200, 'Observed\ndifference', bbox={'facecolor':'white'})
ax.set_xlabel('Conversion rate (percent)')
ax.set_ylabel('Frequency')

plt.tight_layout()
plt.show()
````

**Interpretasi:** Histogram menunjukkan distribusi selisih konversi (dalam %) hasil permutasi, dengan garis pada selisih observasi (0,0368%). Selisih observasi berada **di tengah distribusi**, yang berarti perbedaan sekecil ini sangat mudah terjadi secara kebetulan.

### Menghitung P-Value
Jika `np.mean` diterapkan pada list boolean, hasilnya adalah proporsi nilai `True`. Ini menjadi cara cepat menghitung p-value dari hasil permutasi.

````python
print(np.mean([diff > obs_pct_diff for diff in perm_diffs]))
````

**Interpretasi:** p-value ≈ **0,332**. Sekitar 33% permutasi acak menghasilkan selisih yang sama besar atau lebih besar. Hasil ini **jauh dari signifikan**, sehingga kita **tidak bisa** menyimpulkan harga A lebih baik dari harga B.

Cara lain adalah memakai pendekatan rumus, yaitu chi-square test untuk tabel 2×2 (konversi vs tidak konversi):

````python
survivors = np.array([[200, 23739 - 200], [182, 22588 - 182]])
chi2, p_value, df, _ = stats.chi2_contingency(survivors)

print(f'p-value for single sided test: {p_value / 2:.4f}')
````

**Interpretasi:** p-value satu sisi dari chi-square test ≈ **0,35**, mirip dengan hasil permutation test (0,33). Ini menunjukkan bahwa pendekatan resampling dan pendekatan rumus memberikan kesimpulan yang sama.

## 3.5 t-Tests
**Teori:** Sebelum ada komputer, permutation test sulit dilakukan, sehingga digunakan pendekatan berbasis rumus. Salah satunya **t-test**, yang menggunakan distribusi t (Chapter 2):
$$t = \frac{\bar{x}_B - \bar{x}_A}{\sqrt{\frac{s_A^2}{n_A} + \frac{s_B^2}{n_B}}}$$
Opsi `equal_var=False` menjalankan **Welch's t-test**, yang tidak mengasumsikan variance kedua kelompok sama.

Kita uji ulang data waktu sesi Page A vs Page B:

````python
res = stats.ttest_ind(session_times[session_times.Page == 'Page A'].Time, 
                      session_times[session_times.Page == 'Page B'].Time,
                      equal_var=False)
print(f'p-value for single sided test: {res.pvalue / 2:.4f}')
````

**Interpretasi:** p-value satu sisi dari Welch's t-test adalah **0,1408**, cukup dekat dengan hasil permutation test (≈0,12). Kesimpulannya sama: perbedaan waktu sesi Page A dan Page B **tidak signifikan** pada α = 0,05.

````python
tstat, pvalue, df = sm.stats.ttest_ind(
    session_times[session_times.Page == 'Page A'].Time, 
    session_times[session_times.Page == 'Page B'].Time,
    usevar='unequal', alternative='smaller')
print(f'p-value: {pvalue:.4f}')
````

**Interpretasi:** Fungsi `ttest_ind` dari `statsmodels` dengan `alternative='smaller'` memberikan p-value yang sama (0,1408), karena ini memang uji satu sisi yang sama dengan perhitungan scipy di atas.

## 3.6 Multiple Testing
**Teori (tanpa kode):** Jika kita melakukan **banyak uji sekaligus**, peluang mendapatkan setidaknya satu hasil "signifikan" secara kebetulan meningkat drastis. Dengan 20 uji dan α = 0,05, peluang minimal satu *false positive* adalah $1 - 0{,}95^{20} \approx 64\%$.

Solusinya:
- **Koreksi Bonferroni**: gunakan α / jumlah uji.
- **False Discovery Rate (FDR)**: kendalikan proporsi temuan yang salah.
- Di machine learning, risiko *overfitting* karena terlalu banyak mencoba diatasi dengan **holdout / validation set**.

## 3.7 Degrees of Freedom
**Teori (tanpa kode):** Degrees of freedom (d.f.) adalah **jumlah nilai yang bebas bervariasi**. Jika mean dari 10 data sudah diketahui, hanya 9 data yang bebas karena data ke-10 sudah "ditentukan" oleh mean tersebut, sehingga d.f. = $n - 1$. D.f. memengaruhi bentuk distribusi t, chi-square, dan F.

Di data science, d.f. relevan dalam **regresi dengan variabel kategorikal**. Jika ada $k$ kategori, hanya dibutuhkan $k-1$ dummy variable. Menambahkan dummy ke-$k$ akan menimbulkan **multikolinearitas** (Chapter 4).

## 3.8 ANOVA (Analysis of Variance)
**Teori:** Bagaimana jika kelompoknya lebih dari dua (A, B, C, D)? Membandingkan setiap pasangan akan memunculkan masalah multiple testing. **ANOVA** menguji semuanya sekaligus:
- $H_0$: semua kelompok memiliki mean yang sama.
- Ide utamanya adalah membandingkan **variabilitas antar mean kelompok** dengan **variabilitas di dalam kelompok**.

**Versi permutasi:**
1. Gabungkan semua data.
2. Acak dan bagi kembali menjadi 4 kelompok.
3. Hitung **variance dari 4 mean kelompok**.
4. Ulangi berkali-kali, lalu bandingkan hasilnya dengan variance yang teramati.

Data yang dipakai adalah `four_sessions.csv`, berisi waktu sesi untuk 4 halaman web dengan 5 pengunjung per halaman.

````python
four_sessions = pd.read_csv(FOUR_SESSIONS_CSV)

ax = four_sessions.boxplot(by='Page', column='Time',
                           figsize=(4, 4))
ax.set_xlabel('Page')
ax.set_ylabel('Time (in seconds)')
plt.suptitle('')
plt.title('')

plt.tight_layout()
plt.show()
````

**Interpretasi:** Boxplot keempat halaman menunjukkan **Page 2** memiliki waktu sesi tertinggi dan **Page 4** terendah, tetapi sebarannya saling tumpang tindih.

````python
print(pd.read_csv(FOUR_SESSIONS_CSV).head())
````

````python
observed_variance = four_sessions.groupby('Page').mean().var().iloc[0]
print('Observed means:', four_sessions.groupby('Page').mean().values.ravel())
print('Variance:', observed_variance)
# Permutation test example with stickiness
def perm_test(df):
    df = df.copy()
    df['Time'] = np.random.permutation(df['Time'].values)
    return df.groupby('Page').mean().var().iloc[0]
    
print(perm_test(four_sessions))
````

**Interpretasi:** Mean keempat halaman adalah 172,8; 182,6; 175,6; dan 164,6, dengan **variance antar mean = 55,43**. Fungsi `perm_test` mengacak data dan menghitung variance mean setelah diacak (contoh satu kali permutasi: 19,75).

````python
random.seed(1)
perm_variance = [perm_test(four_sessions) for _ in range(3000)]
print('Pr(Prob)', np.mean([var > observed_variance for var in perm_variance]))

fig, ax = plt.subplots(figsize=(5, 5))
ax.hist(perm_variance, bins=11, rwidth=0.9)
ax.axvline(x = observed_variance, color='black', lw=2)
ax.text(60, 200, 'Observed\nvariance', bbox={'facecolor':'white'})
ax.set_xlabel('Variance')
ax.set_ylabel('Frequency')

plt.tight_layout()
plt.show()
````

**Interpretasi:** Dari 3.000 permutasi, sekitar **7,4%** menghasilkan variance yang lebih besar dari observasi (**p ≈ 0,074**). Karena p > 0,05, perbedaan antar halaman **tidak signifikan** pada α = 0,05, meskipun hasilnya cukup dekat dengan batas.

### F-Statistic
**Teori:** Versi ANOVA berbasis rumus menggunakan **statistik F**:
$$F = \frac{\text{Mean Square Between (variabilitas antar kelompok)}}{\text{Mean Square Within (variabilitas di dalam kelompok / residual)}}$$
Jika $H_0$ benar, F mengikuti distribusi F. Tabel ANOVA dapat dihitung dengan `statsmodels`:

````python
model = smf.ols('Time ~ Page', data=four_sessions).fit()
                
aov_table = sm.stats.anova_lm(model)
print(aov_table)
````

**Interpretasi:** Tabel ANOVA menunjukkan **F = 2,74** dengan d.f. 3 (antar kelompok) dan 16 (residual), serta **p-value = 0,0776**. Hasil ini sangat mirip dengan permutation test (0,074): perbedaan antar halaman belum signifikan pada α = 0,05.

````python
res = stats.f_oneway(four_sessions[four_sessions.Page == 'Page 1'].Time,
                     four_sessions[four_sessions.Page == 'Page 2'].Time,
                     four_sessions[four_sessions.Page == 'Page 3'].Time,
                     four_sessions[four_sessions.Page == 'Page 4'].Time)
print(f'F-Statistic: {res.statistic:.4f}')
print(f'p-value: {res.pvalue:.4f}')
````

**Interpretasi:** `scipy.stats.f_oneway` menghasilkan F dan p-value yang **sama persis** dengan tabel ANOVA `statsmodels` (F ≈ 2,74, p ≈ 0,078).

*Catatan reproduksi:* notebook penulis membagi F dan p-value dengan 2, sehingga muncul F = 1,37 dan p = 0,039. Pembagian F dengan 2 tidak memiliki dasar statistik, dan uji F pada dasarnya memang satu arah. Karena itu di notebook ini nilainya ditampilkan apa adanya agar konsisten dengan tabel ANOVA.

## 3.9 Chi-Square Test
**Teori:** Chi-square test dipakai untuk **data hitungan (count)** pada tabel $r \times c$, misalnya klik vs tidak klik untuk 3 judul iklan.
- $H_0$: proporsi klik sama untuk semua judul.
- Hitung **expected count** di bawah $H_0$, yaitu proporsi klik total dikalikan jumlah tayangan.
- Hitung **Pearson residual**: $R = \frac{O - E}{\sqrt{E}}$
- Statistik chi-square adalah jumlah residual kuadrat: $\chi^2 = \sum R^2$

### Chi-Square Test: Pendekatan Resampling
Data `click_rates.csv` berisi jumlah klik dan no-klik untuk Headline A, B, dan C (masing-masing 1.000 tayangan).

````python
# Table 3-4
click_rate = pd.read_csv(CLICK_RATE_CSV)
clicks = click_rate.pivot(index='Click', columns='Headline', values='Rate')
print(clicks)
````

**Interpretasi:** Headline A mendapat 14 klik, B 8 klik, dan C 12 klik, masing-masing dari 1.000 tayangan.

````python
# Table 3-5
row_average = clicks.mean(axis=1)
pd.DataFrame({
    'Headline A': row_average,
    'Headline B': row_average,
    'Headline C': row_average,
})
````

**Interpretasi:** Jika $H_0$ benar (semua headline sama baiknya), setiap headline **diharapkan** mendapat 34/3 ≈ **11,33 klik** dan 988,67 no-klik. Inilah *expected count*.

````python
# Resampling approach
box = [1] * 34
box.extend([0] * 2966)
random.shuffle(box)

def chi2(observed, expected):
    pearson_residuals = []
    for row, expect in zip(observed, expected):
        pearson_residuals.append([(observe - expect) ** 2 / expect
                                  for observe in row])
    # return sum of squares
    return np.sum(pearson_residuals)

expected_clicks = 34 / 3
expected_noclicks = 1000 - expected_clicks
expected = [expected_clicks, expected_noclicks]
chi2observed = chi2(clicks.values, expected)

def perm_fun(box):
    random.shuffle(box)
    sample_clicks = [sum(box[0:1000]),
                     sum(box[1000:2000]),
                     sum(box[2000:3000])]
    sample_noclicks = [1000 - n for n in sample_clicks]
    return chi2([sample_clicks, sample_noclicks], expected)

perm_chi2 = [perm_fun(box) for _ in range(2000)]

resampled_p_value = sum(perm_chi2 > chi2observed) / len(perm_chi2)
print(f'Observed chi2: {chi2observed:.4f}')
print(f'Resampled p-value: {resampled_p_value:.4f}')
````

**Interpretasi:** Statistik chi-square observasi = **1,666**. Dengan resampling, sekitar **46,6%** hasil acak memiliki chi-square yang lebih besar, sehingga perbedaan klik antar headline **tidak signifikan**.

````python
chisq, pvalue, df, expected = stats.chi2_contingency(clicks)
print(f'Observed chi2: {chisq:.4f}')
print(f'p-value: {pvalue:.4f}')
````

**Interpretasi:** Pendekatan rumus (`chi2_contingency`) dengan d.f. = (2−1)(3−1) = 2 menghasilkan p-value ≈ **0,435**, mirip dengan hasil resampling. Ketiga headline tidak bisa dibedakan secara statistik dengan data ini.

Algoritma di atas mengambil sampel **tanpa pengembalian**. Kita juga bisa mengambil sampel **dengan pengembalian** (*bootstrap*), dan hasilnya serupa:

````python
expected = [expected_clicks, expected_noclicks]
def sample_with_replacement(box):
    sample_clicks = [sum(random.sample(box, 1000)),
                     sum(random.sample(box, 1000)),
                     sum(random.sample(box, 1000))]
    sample_noclicks = [1000 - n for n in sample_clicks]
    return chi2([sample_clicks, sample_noclicks], expected)

perm_chi2 = [sample_with_replacement(box) for _ in range(2000)]

resampled_p_value = sum(perm_chi2 > chi2observed) / len(perm_chi2)
print(f'Observed chi2: {chi2observed:.4f}')
print(f'Resampled p-value: {resampled_p_value:.4f}')
````

**Interpretasi:** Resampling dengan pengembalian memberi p-value ≈ 0,48, juga konsisten. Kesimpulannya tetap: tidak ada headline yang terbukti lebih baik.

### Distribusi Chi-Square
Grafik berikut menunjukkan kurva distribusi chi-square untuk beberapa degrees of freedom. Semakin besar d.f., kurvanya semakin bergeser ke kanan dan melebar. Untuk tabel $r \times c$, d.f. $= (r-1)(c-1)$.

````python
x = [1 + i * (30 - 1) / 99 for i in range(100)]

chi = pd.DataFrame({
    'x': x,
    'chi_1': stats.chi2.pdf(x, df=1),
    'chi_2': stats.chi2.pdf(x, df=2),
    'chi_5': stats.chi2.pdf(x, df=5),
    'chi_10': stats.chi2.pdf(x, df=10),
    'chi_20': stats.chi2.pdf(x, df=20),
})
fig, ax = plt.subplots(figsize=(4, 2.5))
ax.plot(chi.x, chi.chi_1, color='black', linestyle='-', label='1')
ax.plot(chi.x, chi.chi_2, color='black', linestyle=(0, (1, 1)), label='2')
ax.plot(chi.x, chi.chi_5, color='black', linestyle=(0, (2, 1)), label='5')
ax.plot(chi.x, chi.chi_10, color='black', linestyle=(0, (3, 1)), label='10')
ax.plot(chi.x, chi.chi_20, color='black', linestyle=(0, (4, 1)), label='20')
ax.legend(title='df')

plt.tight_layout()
plt.show()
````

## 3.10 Multi-Arm Bandit Algorithm
**Teori (tanpa kode):** A/B test klasik punya kelemahan: kita harus menunggu sampai eksperimen selesai, dan selama itu setengah pengunjung tetap diberi versi yang lebih buruk. **Multi-arm bandit** (nama dari mesin slot dengan banyak tuas) menyeimbangkan dua hal:
- **Exploration**: terus mencoba semua opsi untuk belajar.
- **Exploitation**: lebih sering memakai opsi yang sejauh ini terbaik.

Contoh algoritma **epsilon-greedy**: dengan probabilitas ε (misalnya 5%), pilih opsi secara acak (eksplorasi). Selebihnya, pilih opsi dengan hasil terbaik saat ini (eksploitasi). Algoritma lain yang lebih canggih adalah **Thompson sampling** (pendekatan Bayesian). Pendekatan bandit cocok untuk optimasi web yang terus berjalan.

## 3.11 Power and Sample Size
**Teori:**
- **Effect size**: besar efek minimum yang ingin bisa dideteksi (misalnya kenaikan klik dari 1,1% ke 1,21%).
- **Power**: probabilitas mendeteksi efek tersebut **jika efeknya memang ada**, atau $1 - P(\text{Type II error})$. Target yang umum adalah 80%.
- **Significance level (α)**: biasanya 0,05.
- **Sample size**: jumlah data yang dibutuhkan.

Keempat besaran ini saling terkait. Jika tiga di antaranya ditentukan, besaran keempat bisa dihitung. Di `statsmodels`, `proportion_effectsize` mengubah selisih proporsi menjadi effect size (Cohen's h), lalu `TTestIndPower().solve_power()` menghitung ukuran sampel.

**Kasus 1:** CTR saat ini 1,1%, dan kita ingin mendeteksi kenaikan 10% (menjadi 1,21%):

````python
effect_size = sm.stats.proportion_effectsize(0.0121, 0.011)
analysis = sm.stats.TTestIndPower()
result = analysis.solve_power(effect_size=effect_size, 
                              alpha=0.05, power=0.8, alternative='larger')
print('Sample Size: %.3f' % result)
````

**Interpretasi:** Untuk mendeteksi kenaikan CTR dari 1,1% ke 1,21% (kenaikan relatif 10%) dengan power 80% dan α = 0,05, dibutuhkan sekitar **116.602 tayangan per kelompok**. Efek kecil membutuhkan sampel yang sangat besar.

**Kasus 2:** bagaimana jika kita ingin mendeteksi kenaikan yang lebih besar, yaitu 50% (dari 1,1% ke 1,65%)?

````python
effect_size = sm.stats.proportion_effectsize(0.0165, 0.011)
analysis = sm.stats.TTestIndPower()
result = analysis.solve_power(effect_size=effect_size, 
                              alpha=0.05, power=0.8, alternative='larger')
print('Sample Size: %.3f' % result)
````

**Interpretasi:** Untuk efek yang lebih besar (1,1% → 1,65%), cukup sekitar **5.488 tayangan per kelompok**, sekitar 21 kali lebih sedikit dari kasus 1. Pelajarannya: **semakin kecil efek yang ingin dideteksi, semakin besar sampel yang dibutuhkan**. Power analysis sebaiknya dilakukan **sebelum** eksperimen dimulai.

## Ringkasan Chapter 3
1. **A/B testing** membandingkan dua treatment dengan **randomisasi**, sehingga perbedaan hasil hanya bisa disebabkan treatment atau kebetulan.
2. **Hypothesis testing** melindungi kita dari tertipu kebetulan. Kita mengasumsikan $H_0$ (tidak ada efek), lalu melihat apakah data cukup ekstrem untuk menolaknya.
3. **Permutation test** (resampling tanpa pengembalian) menguji hipotesis tanpa asumsi distribusi, dan sangat intuitif.
4. **p-value** adalah peluang memperoleh hasil seekstrem observasi jika $H_0$ benar. α = 0,05 adalah konvensi, bukan hukum. Waspadai **Type I** dan **Type II error**.
5. **t-test** adalah versi rumus untuk membandingkan dua mean. **ANOVA** dan **F-statistic** dipakai untuk lebih dari dua kelompok. **Chi-square test** dipakai untuk data hitungan dalam tabel.
6. **Multiple testing** meningkatkan risiko false positive. Atasi dengan koreksi Bonferroni, FDR, atau holdout set.
7. **Degrees of freedom** memengaruhi bentuk distribusi uji, dan berkaitan dengan aturan $k-1$ dummy variable di regresi.
8. **Multi-arm bandit** menyeimbangkan eksplorasi dan eksploitasi, sehingga lebih efisien daripada A/B test klasik untuk optimasi berkelanjutan.
9. **Power analysis** menentukan ukuran sampel yang dibutuhkan. Efek yang kecil membutuhkan sampel yang **sangat besar**.

**Kaitan dengan Machine Learning:** konsep ini dipakai untuk mengevaluasi apakah model baru **benar-benar lebih baik** dari model lama (bukan sekadar kebetulan), untuk A/B testing model di produksi, dan untuk memahami risiko *overfitting* akibat terlalu banyak mencoba (multiple testing).
