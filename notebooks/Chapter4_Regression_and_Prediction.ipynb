## 4.1 Simple Linear Regression
### The Regression Equation
**Teori:** Simple linear regression memodelkan hubungan **satu prediktor** $X$ dengan outcome $Y$:
$$Y = b_0 + b_1 X + e$$
- $b_0$ = **intercept**: nilai Y ketika X = 0.
- $b_1$ = **slope / koefisien**: perubahan Y untuk setiap kenaikan 1 unit X.
- $e$ = **error / residual**.

Berbeda dengan korelasi (yang hanya mengukur kekuatan hubungan), regresi **mengkuantifikasi bentuk** hubungan tersebut dan bisa dipakai untuk prediksi.

**Data:** `LungDisease.csv` berisi `Exposure` (lama paparan debu kapas, dalam tahun) dan `PEFR` (*peak expiratory flow rate*, ukuran kapasitas paru-paru) dari pekerja pabrik kapas.

```python
lung = pd.read_csv(LUNG_CSV)

lung.plot.scatter(x='Exposure', y='PEFR')

plt.tight_layout()
plt.show()
```

**Interpretasi:** Scatterplot menunjukkan tren **menurun**: semakin lama pekerja terpapar debu kapas, semakin rendah PEFR-nya (fungsi paru-paru memburuk).

Fitting model dengan `LinearRegression` dari *scikit-learn*:

```python
predictors = ['Exposure']
outcome = 'PEFR'

model = LinearRegression()
model.fit(lung[predictors], lung[outcome])

print(f'Intercept: {model.intercept_:.3f}')
print(f'Coefficient Exposure: {model.coef_[0]:.3f}')
```

**Interpretasi:** Persamaan regresinya adalah **PEFR = 424,58 − 4,185 × Exposure**.
- Intercept 424,58: perkiraan PEFR pekerja yang **belum pernah terpapar**.
- Slope −4,185: setiap **tambahan 1 tahun paparan**, PEFR turun sekitar **4,19 unit**.

```python
fig, ax = plt.subplots(figsize=(4, 4))
ax.set_xlim(0, 23)
ax.set_ylim(295, 450)
ax.set_xlabel('Exposure')
ax.set_ylabel('PEFR')
ax.plot((0, 23), model.predict(pd.DataFrame({'Exposure': [0, 23]})))
ax.text(0.4, model.intercept_, r'$b_0$', size='larger')

x = pd.DataFrame({'Exposure': [7.5,17.5]})
y = model.predict(x)
ax.plot((7.5, 7.5, 17.5), (y[0], y[1], y[1]), '--')
ax.text(5, np.mean(y), r'$\Delta Y$', size='larger')
ax.text(12, y[1] - 10, r'$\Delta X$', size='larger')
ax.text(12, 390, r'$b_1 = \frac{\Delta Y}{\Delta X}$', size='larger')

plt.tight_layout()
plt.show()
```

**Interpretasi:** Grafik ini menampilkan garis regresi beserta anotasi intercept ($b_0$) dan slope ($b_1$), yaitu perubahan Y untuk setiap kenaikan 1 unit X.

### Fitted Values and Residuals
**Teori:**
- **Fitted value** ($\hat{Y}_i$): nilai prediksi model untuk data ke-$i$, yaitu $\hat{Y}_i = \hat{b}_0 + \hat{b}_1 X_i$.
- **Residual** ($\hat{e}_i$): selisih nilai asli dengan prediksi, yaitu $\hat{e}_i = Y_i - \hat{Y}_i$.

### Least Squares
Koefisien dicari dengan metode **ordinary least squares (OLS)**, yaitu meminimalkan **Residual Sum of Squares**:
$$RSS = \sum_{i=1}^{n} (Y_i - \hat{b}_0 - \hat{b}_1 X_i)^2$$
OLS punya solusi rumus tertutup sehingga cepat dihitung, tetapi karena memakai kuadrat, metode ini **sensitif terhadap outlier**.

Method `predict` dipakai untuk menghitung fitted values:

```python
fitted = model.predict(lung[predictors])
residuals = lung[outcome] - fitted
```

```python
ax = lung.plot.scatter(x='Exposure', y='PEFR', figsize=(4, 4))
ax.plot(lung.Exposure, fitted)
for x, yactual, yfitted in zip(lung.Exposure, lung.PEFR, fitted): 
    ax.plot((x, x), (yactual, yfitted), '--', color='C1')

plt.tight_layout()
plt.show()
```

**Interpretasi:** Garis vertikal dari setiap titik ke garis regresi adalah **residual**. Metode least squares mencari garis yang membuat total kuadrat panjang garis-garis vertikal ini sekecil mungkin.

## 4.2 Multiple Linear Regression
**Teori:** Dengan banyak prediktor:
$$Y = b_0 + b_1 X_1 + b_2 X_2 + \dots + b_p X_p + e$$
Setiap koefisien $b_j$ diinterpretasikan sebagai perubahan Y untuk kenaikan 1 unit $X_j$, **dengan asumsi prediktor lain tetap (dikontrol)**.

**Data:** `house_sales.csv` berisi data penjualan rumah di King County. File ini **dipisahkan dengan tab**, sehingga perlu `sep='\t'`. Outcome-nya adalah `AdjSalePrice` (harga jual yang disesuaikan inflasi).

```python
subset = ['AdjSalePrice', 'SqFtTotLiving', 'SqFtLot', 'Bathrooms', 
          'Bedrooms', 'BldgGrade']

house = pd.read_csv(HOUSE_CSV, sep='\t')
print(house[subset].head())
```

```python
predictors = ['SqFtTotLiving', 'SqFtLot', 'Bathrooms', 
              'Bedrooms', 'BldgGrade']
outcome = 'AdjSalePrice'

house_lm = LinearRegression()
house_lm.fit(house[predictors], house[outcome])

print(f'Intercept: {house_lm.intercept_:.3f}')
print('Coefficients:')
for name, coef in zip(predictors, house_lm.coef_):
    print(f' {name}: {coef}')
```

**Interpretasi koefisien** (dengan asumsi variabel lain tetap):
- `SqFtTotLiving` ≈ **+228,8**: setiap tambahan 1 sqft luas bangunan menaikkan harga sekitar \$229.
- `BldgGrade` ≈ **+106.107**: kenaikan 1 tingkat kualitas bangunan menaikkan harga sekitar \$106 ribu.
- `Bedrooms` ≈ **−47.770** dan `Bathrooms` ≈ −19.443 bernilai **negatif**, terlihat aneh. Penjelasannya ada di bagian *Correlated Predictors* (4.5).
- `SqFtLot` ≈ −0,06: luas tanah hampir tidak berpengaruh.

### Assessing the Model
**Teori:**
- **RMSE (Root Mean Squared Error)**: $\sqrt{\frac{1}{n}\sum (y_i - \hat{y}_i)^2}$. Ini adalah rata-rata besar kesalahan prediksi dalam satuan Y, dan merupakan metrik paling penting untuk prediksi.
- **RSE (Residual Standard Error)**: mirip RMSE tetapi dibagi $(n - p - 1)$ (degrees of freedom).
- **R² (coefficient of determination)**: proporsi variasi Y yang dijelaskan model, bernilai 0 sampai 1.
$$R^2 = 1 - \frac{\sum (y_i - \hat{y}_i)^2}{\sum (y_i - \bar{y})^2}$$
- **t-statistic** dan **p-value** tiap koefisien: menguji apakah koefisien berbeda signifikan dari nol, dengan $t = \hat{b}/SE(\hat{b})$.

```python
fitted = house_lm.predict(house[predictors])
RMSE = np.sqrt(mean_squared_error(house[outcome], fitted))
r2 = r2_score(house[outcome], fitted)
print(f'RMSE: {RMSE:.0f}')
print(f'r2: {r2:.4f}')
```

**Interpretasi:** **RMSE ≈ \$261.220**, artinya rata-rata kesalahan prediksi harga sekitar \$261 ribu. **R² = 0,54**, artinya model menjelaskan sekitar 54% variasi harga rumah. Sisanya dipengaruhi faktor lain yang belum dimasukkan, terutama **lokasi**.

*scikit-learn* fokus pada prediksi, sedangkan *statsmodels* memberikan analisis statistik yang lebih lengkap (SE, t-stat, p-value, CI tiap koefisien):

```python
model = sm.OLS(house[outcome], house[predictors].assign(const=1))
results = model.fit()
print(results.summary())
```

**Interpretasi:** Ringkasan `statsmodels` memberi informasi yang lebih lengkap: R² = 0,541, Adjusted R² = 0,540, serta **t-statistic** dan **p-value** setiap koefisien. Hampir semua prediktor memiliki p-value ≈ 0 (sangat signifikan) karena jumlah data sangat besar (±22 ribu rumah).

### Model Selection and Stepwise Regression
**Teori:** Menambah prediktor selalu menaikkan R² di data training, tetapi belum tentu meningkatkan prediksi pada data baru (*overfitting*). Prinsip **Occam's razor**: pilih model yang paling sederhana jika performanya setara.

Metrik yang **memberi penalti** pada jumlah variabel:
- **Adjusted R²**: $R^2_{adj} = 1 - (1-R^2)\frac{n-1}{n-p-1}$
- **AIC (Akaike's Information Criterion)**: $AIC = 2P + n\log(RSS/n)$. Semakin kecil, semakin baik.
- **BIC**: mirip AIC dengan penalti yang lebih besar.

Metode pemilihan variabel:
- **All subset regression**: coba semua kombinasi. Akurat, tetapi lambat.
- **Forward selection**: mulai dari model kosong, lalu tambahkan variabel satu per satu.
- **Backward elimination**: mulai dari model penuh, lalu buang variabel satu per satu.
- **Stepwise**: kombinasi forward dan backward.
- **Regularization** (Ridge, Lasso): koefisien "dikecilkan" alih-alih dibuang.

Pertama, kita fit model dengan lebih banyak prediktor, termasuk variabel kategorikal `PropertyType` yang diubah menjadi dummy:

```python
predictors = ['SqFtTotLiving', 'SqFtLot', 'Bathrooms', 'Bedrooms',
              'BldgGrade', 'PropertyType', 'NbrLivingUnits',
              'SqFtFinBasement', 'YrBuilt', 'YrRenovated', 
              'NewConstruction']

X = pd.get_dummies(house[predictors], drop_first=True, dtype=int)
X['NewConstruction'] = [1 if nc else 0 for nc in X['NewConstruction']]

house_full = sm.OLS(house[outcome], X.assign(const=1))
results = house_full.fit()
print(results.summary())
```

**Interpretasi:** Dengan menambahkan lebih banyak prediktor (tipe properti, tahun dibangun, basement, dan lain-lain), R² naik menjadi **0,595**. Namun, apakah semua variabel tambahan itu benar-benar diperlukan? Stepwise regression akan membantu menjawabnya.

Stepwise selection menggunakan fungsi `stepwise_selection` dari paket **dmba** dengan AIC sebagai skor:

```python
y = house[outcome]

def train_model(variables):
    if len(variables) == 0:
        return None
    model = LinearRegression()
    model.fit(X[variables], y)
    return model

def score_model(model, variables):
    if len(variables) == 0:
        return AIC_score(y, [y.mean()] * len(y), model, df=1)
    return AIC_score(y, model.predict(X[variables]), model)

best_model, best_variables = stepwise_selection(X.columns, train_model, score_model, 
                                                verbose=True)

print()
print(f'Intercept: {best_model.intercept_:.3f}')
print('Coefficients:')
for name, coef in zip(best_variables, best_model.coef_):
    print(f' {name}: {coef}')
```

**Interpretasi:** Stepwise memulai dari model konstan, lalu **menambahkan atau membuang variabel satu per satu** selama skor AIC masih turun. Variabel pertama yang masuk adalah `SqFtTotLiving`, kemudian `BldgGrade`, `YrBuilt`, dan seterusnya. Variabel yang tidak menurunkan AIC (misalnya `NbrLivingUnits`, `YrRenovated`, `NewConstruction`) **tidak dipakai**. Hasilnya adalah model yang lebih sederhana tetapi tetap akurat.

### Weighted Regression
**Teori:** Pada **weighted least squares**, setiap record diberi bobot berbeda saat fitting. Kegunaannya:
- Data dengan variance berbeda (*heteroskedasticity*): beri bobot lebih kecil pada data yang lebih "berisik".
- Data agregat: bobot sesuai jumlah observasi yang diwakili.
- Memberi bobot lebih besar pada data yang lebih **relevan** atau **baru**.

Di sini penjualan yang lebih baru diberi bobot lebih besar: bobot = tahun penjualan − 2005. Kolom tahun diambil dari `DocumentDate`:

```python
house['Year'] = [int(date.split('-')[0]) for date in house.DocumentDate]
house['Year'] = house.DocumentDate.apply(lambda d: int(d.split('-')[0]))
house['Weight'] = house.Year - 2005
```

```python
predictors = ['SqFtTotLiving', 'SqFtLot', 'Bathrooms', 
              'Bedrooms', 'BldgGrade']
outcome = 'AdjSalePrice'

house_wt = LinearRegression()
house_wt.fit(house[predictors], house[outcome], sample_weight=house.Weight)

pd.concat([
    pd.DataFrame({
        'predictor': predictors,
        'house_lm': house_lm.coef_,
        'house_wt': house_wt.coef_,    
    }),
    pd.DataFrame({
        'predictor': ['intercept'],
        'house_lm': house_lm.intercept_,
        'house_wt': house_wt.intercept_,    
    })
])
```

**Interpretasi:** Koefisien model berbobot (`house_wt`) sedikit berbeda dari model biasa (`house_lm`). Misalnya, `SqFtTotLiving` naik dari 228,8 menjadi 245,0. Artinya, pada penjualan yang lebih baru, nilai per sqft rumah cenderung lebih tinggi.

```python
residuals = pd.DataFrame({
    'abs_residual_lm': np.abs(house_lm.predict(house[predictors]) - house[outcome]),
    'abs_residual_wt': np.abs(house_wt.predict(house[predictors]) - house[outcome]),
    'Year': house['Year'],
})
print(residuals.head())
# axes = residuals.boxplot(['abs_residual_lm', 'abs_residual_wt'], by='Year', figsize=(10, 4))
# axes[0].set_ylim(0, 300000)

pd.DataFrame(([year, np.mean(group['abs_residual_lm']), np.mean(group['abs_residual_wt'])] 
              for year, group in residuals.groupby('Year')),
             columns=['Year', 'mean abs_residual_lm', 'mean abs_residual_wt'])
# for year, group in residuals.groupby('Year'):
#     print(year, np.mean(group['abs_residual_lm']), np.mean(group['abs_residual_wt']))
```

## 4.3 Prediction Using Regression
**Teori (tanpa kode di notebook penulis):**
- **Bahaya ekstrapolasi**: model regresi hanya valid di dalam **rentang data** yang dipakai untuk membangunnya. Contohnya, model harga rumah tidak bisa dipakai untuk memprediksi harga tanah kosong (luas bangunan = 0).
- **Confidence interval**: ketidakpastian di sekitar **koefisien** atau **rata-rata prediksi**.
- **Prediction interval**: ketidakpastian untuk **satu nilai prediksi individual**. Interval ini selalu **lebih lebar** karena mencakup ketidakpastian model **ditambah** variasi individual (error $e$).

Bootstrap juga bisa dipakai untuk membuat interval ini. Caranya, resample baris data, fit ulang model, lalu catat prediksinya. Untuk prediction interval, tambahkan residual yang diambil secara acak.

## 4.4 Factor Variables in Regression
### Dummy Variables Representation
**Teori:** Regresi membutuhkan input numerik, sehingga variabel kategorikal (*factor*) harus diubah menjadi variabel biner 0/1:
- **One-hot encoding**: satu kolom untuk setiap kategori. Cocok untuk banyak algoritma ML (misalnya tree, neural net).
- **Reference coding**: $k$ kategori diwakili $k-1$ dummy (`drop_first=True`). Kategori yang dibuang menjadi **referensi (baseline)**. Ini **wajib** untuk regresi linear dengan intercept, karena jika semua $k$ dummy dimasukkan, kolom-kolomnya menjadi dependen linear (*multicollinearity*).

Variabel `PropertyType` memiliki 3 kategori: Multiplex, Single Family, dan Townhouse.

```python
print(house.PropertyType.head())
```

```python
print(pd.get_dummies(house['PropertyType']).head(6))
```

**Interpretasi:** `pd.get_dummies` menghasilkan **one-hot encoding**: 3 kolom, satu untuk setiap tipe properti.

```python
print(pd.get_dummies(house['PropertyType'], drop_first=True).head(6))
```

**Interpretasi:** Dengan `drop_first=True`, kolom **Multiplex dibuang** dan menjadi kategori referensi. Rumah Multiplex ditandai dengan nilai False pada kedua kolom lainnya. Inilah **reference coding** ($k-1$ dummy).

```python
predictors = ['SqFtTotLiving', 'SqFtLot', 'Bathrooms', 'Bedrooms',
              'BldgGrade', 'PropertyType']

X = pd.get_dummies(house[predictors], drop_first=True)

house_lm_factor = LinearRegression()
house_lm_factor.fit(X, house[outcome])

print(f'Intercept: {house_lm_factor.intercept_:.3f}')
print('Coefficients:')
for name, coef in zip(X.columns, house_lm_factor.coef_):
    print(f' {name}: {coef}')
```

**Interpretasi:** Koefisien `PropertyType_Single Family` ≈ **−84.678** dan `PropertyType_Townhouse` ≈ **−115.122**. Artinya, **dibandingkan Multiplex** (referensi) dan dengan variabel lain tetap, rumah Single Family diperkirakan \$85 ribu lebih murah, dan Townhouse \$115 ribu lebih murah.

### Factor Variables with Many Levels
**Teori:** `ZipCode` memiliki puluhan kategori. Membuat dummy untuk setiap kode pos akan menghasilkan terlalu banyak variabel, padahal beberapa kode pos hanya punya sangat sedikit data. Solusinya adalah **mengelompokkan** kategori.

Strategi dari buku: fit model awal, hitung **median residual** per kode pos, urutkan, lalu bagi menjadi **5 grup (ZipGroup)** dengan jumlah penjualan yang kurang lebih sama. Kode pos dengan residual positif (harga lebih tinggi dari prediksi) mewakili lokasi yang lebih mahal.

```python
print(pd.DataFrame(house['ZipCode'].value_counts()).transpose())
```

**Interpretasi:** Jumlah penjualan per kode pos sangat bervariasi. Ada kode pos dengan ratusan penjualan (98038: 788), tetapi ada juga yang hanya 1 penjualan. Membuat dummy untuk setiap kode pos tidak efisien.

```python
house = pd.read_csv(HOUSE_CSV, sep='\t')

predictors = ['SqFtTotLiving', 'SqFtLot', 'Bathrooms', 
              'Bedrooms', 'BldgGrade']
outcome = 'AdjSalePrice'

house_lm = LinearRegression()
house_lm.fit(house[predictors], house[outcome])


zip_residuals = pd.DataFrame({
    'ZipCode': house['ZipCode'],
    'residual' : house[outcome] - house_lm.predict(house[predictors]),
})

zip_groups = pd.DataFrame([
    {
        'ZipCode': zipCode,
        'count': len(x),
        'median_residual': x.residual.median()
    } 
    for zipCode, x in zip_residuals.groupby('ZipCode')
]).sort_values('median_residual')

zip_groups['cum_count'] = np.cumsum(zip_groups['count'])
zip_groups['ZipGroup'] = pd.qcut(zip_groups['cum_count'], 5, labels=False, retbins=False)
zip_groups.head()
print(zip_groups.ZipGroup.value_counts().sort_index())
```

**Interpretasi:** Kode pos berhasil dikelompokkan menjadi **5 ZipGroup**, masing-masing berisi 16 kode pos. Grup 0 mewakili lokasi dengan harga paling rendah dibanding prediksi, dan grup 4 lokasi paling mahal.

```python
to_join = zip_groups[['ZipCode', 'ZipGroup']].set_index('ZipCode')
house = house.join(to_join, on='ZipCode')
house['ZipGroup'] = house['ZipGroup'].astype('category')
```

## 4.5 Interpreting the Regression Equation
### Correlated Predictors
**Teori:** Jika prediktor saling berkorelasi, koefisien regresi menjadi **sulit ditafsirkan** dan bisa bertanda "aneh". Contoh klasik: koefisien `Bedrooms` bernilai **negatif**. Apakah menambah kamar tidur menurunkan harga? Tidak. Jika *luas rumah tetap*, menambah kamar berarti kamarnya menjadi lebih kecil. Informasi `Bedrooms` sudah "tertangkap" oleh `SqFtTotLiving`.

**Multicollinearity** (korelasi yang sangat tinggi antar prediktor) membuat estimasi koefisien tidak stabil. Kasus ekstrem (*perfect multicollinearity*) terjadi jika satu variabel adalah kombinasi linear dari variabel lain.

Hasil stepwise regression sebelumnya:

```python
print(f'Intercept: {best_model.intercept_:.3f}')
print('Coefficients:')
for name, coef in zip(best_variables, best_model.coef_):
    print(f' {name}: {coef}')
```

**Interpretasi:** Koefisien `Bedrooms` tetap **negatif** (≈ −51.947) meskipun `Bathrooms` kini positif. Ini adalah efek **korelasi antar prediktor**: `Bedrooms`, `Bathrooms`, dan `SqFtTotLiving` saling berkorelasi karena rumah besar cenderung punya lebih banyak kamar.

```python
predictors = ['Bedrooms', 'BldgGrade', 'PropertyType', 'YrBuilt']
outcome = 'AdjSalePrice'

X = pd.get_dummies(house[predictors], drop_first=True)

reduced_lm = LinearRegression()
reduced_lm.fit(X, house[outcome])


print(f'Intercept: {reduced_lm.intercept_:.3f}')
print('Coefficients:')
for name, coef in zip(X.columns, reduced_lm.coef_):
    print(f' {name}: {coef}')
```

**Interpretasi:** Jika `SqFtTotLiving`, `Bathrooms`, dan `SqFtFinBasement` **dikeluarkan** dari model, koefisien `Bedrooms` berubah menjadi **positif** (≈ +27.151). Ini membuktikan bahwa tanda negatif sebelumnya muncul karena luas rumah sudah dikontrol. Setelah luas rumah tidak ada di model, `Bedrooms` menjadi "proxy" untuk ukuran rumah.

### Confounding Variables
**Teori:** **Confounding variable** adalah variabel penting yang **tidak dimasukkan** ke dalam model, sehingga koefisien lain menjadi bias. Contoh di sini: **lokasi** (ZipGroup) sangat memengaruhi harga. Tanpa lokasi, model bisa salah menafsirkan efek variabel lain. Kode berikut menambahkan `ZipGroup` sebagai prediktor:

```python
predictors = ['SqFtTotLiving', 'SqFtLot', 'Bathrooms', 'Bedrooms',
              'BldgGrade', 'PropertyType', 'ZipGroup']
outcome = 'AdjSalePrice'

X = pd.get_dummies(house[predictors], drop_first=True)

confounding_lm = LinearRegression()
confounding_lm.fit(X, house[outcome])

print(f'Intercept: {confounding_lm.intercept_:.3f}')
print('Coefficients:')
for name, coef in zip(X.columns, confounding_lm.coef_):
    print(f' {name}: {coef}')
```

**Interpretasi:** Setelah `ZipGroup` dimasukkan, koefisien grup naik secara bertahap (ZipGroup_1 ≈ +53 ribu, ZipGroup_2 ≈ +116 ribu, dan seterusnya, relatif terhadap grup 0). Ini mengonfirmasi bahwa **lokasi adalah faktor yang sangat besar**. Koefisien `SqFtLot` juga berubah dari negatif menjadi positif (+0,45): sebelumnya efek luas tanah "tercampur" dengan efek lokasi, karena rumah di pinggiran kota punya tanah luas tetapi lokasinya lebih murah.

### Interactions and Main Effects
**Teori:**
- **Main effect**: efek satu prediktor secara independen.
- **Interaction**: efek satu prediktor **bergantung pada nilai prediktor lain**. Contohnya, harga per sqft di lokasi mahal jauh lebih tinggi daripada di lokasi murah.

Di formula `statsmodels`, notasi `SqFtTotLiving*ZipGroup` otomatis memasukkan kedua main effect **dan** interaksinya (`SqFtTotLiving:ZipGroup`).

```python
model = smf.ols(formula='AdjSalePrice ~  SqFtTotLiving*ZipGroup + SqFtLot + ' +
     'Bathrooms + Bedrooms + BldgGrade + PropertyType', data=house)
results = model.fit()
print(results.summary())
```

**Interpretasi:** Model interaksi menaikkan R² menjadi **0,682**. Lihat koefisien interaksinya:
- `SqFtTotLiving` (ZipGroup 0, referensi): ≈ **\$115 per sqft**.
- `SqFtTotLiving:ZipGroup[T.4]` ≈ +227, sehingga di grup lokasi termahal nilainya ≈ 115 + 227 = **\$342 per sqft**.

Artinya, **efek luas rumah terhadap harga bergantung pada lokasi**: di lokasi mahal, setiap sqft tambahan bernilai hampir 3 kali lipat dibanding di lokasi murah.

*Catatan:* hasil ini sedikit berbeda dari versi R di buku karena perbedaan cara pengelompokan (*binning*) ZipGroup.

## 4.6 Regression Diagnostics (Testing the Assumptions)
**Teori:** Asumsi regresi linear OLS:
1. Hubungan X dan Y **linear**.
2. Error **independen**.
3. Error memiliki **variance konstan** (*homoskedasticity*).
4. Error **berdistribusi normal** (terutama penting untuk inferensi p-value dan CI).

Untuk **prediksi**, asumsi 3 dan 4 tidak terlalu krusial. Untuk **inferensi** (menafsirkan p-value), asumsi tersebut penting.

### Outliers
**Teori:** Outlier dalam regresi adalah record yang nilai Y-nya **jauh dari prediksi** (residual besar). Ukuran yang dipakai adalah **standardized residual**, yaitu residual dibagi standard error-nya. Nilai di luar ±2,5 atau ±3 dianggap outlier. Outlier bisa berupa kesalahan data, kasus khusus, atau penipuan.

Model regresi khusus untuk kode pos **98105**:

```python
house_98105 = house.loc[house['ZipCode'] == 98105, ]

predictors = ['SqFtTotLiving', 'SqFtLot', 'Bathrooms', 'Bedrooms',
              'BldgGrade']
outcome = 'AdjSalePrice'

house_outlier = sm.OLS(house_98105[outcome], house_98105[predictors].assign(const=1))
result_98105 = house_outlier.fit()
print(result_98105.summary())
```

**Interpretasi:** Untuk kode pos 98105 (313 rumah), R² = **0,795**. Dalam satu lokasi yang homogen, model dapat menjelaskan variasi harga jauh lebih baik.

`OLSInfluence` menghitung standardized residual. Kita cari record dengan residual paling negatif:

```python
influence = OLSInfluence(result_98105)
sresiduals = influence.resid_studentized_internal

print(sresiduals.idxmin(), sresiduals.min())
```

**Interpretasi:** Record dengan indeks **24333** memiliki standardized residual ≈ **−4,33**, jauh di luar batas ±2,5 sehingga tergolong **outlier**. Harganya jauh lebih rendah dari prediksi model.

```python
print(result_98105.resid.loc[sresiduals.idxmin()])
```

**Interpretasi:** Residual mentah untuk rumah tersebut ≈ −\$757.754. Harga aktualnya lebih rendah sekitar \$758 ribu dari prediksi model.

```python
outlier = house_98105.loc[sresiduals.idxmin(), :]
print('AdjSalePrice', outlier[outcome])
print(outlier[predictors])
```

**Interpretasi:** Rumah ini (2.900 sqft, 3 kamar mandi, 6 kamar tidur, grade 7) terjual hanya **\$119.748**. Harga ini sangat tidak wajar untuk lokasi 98105. Menurut buku, setelah ditelusuri, transaksi ini ternyata **hanya penjualan sebagian kepemilikan (partial interest)**, sehingga bukan data penjualan normal. Outlier seperti ini perlu **diinvestigasi**, bukan langsung dibuang atau dipertahankan.

### Influential Values
**Teori:** **Influential observation** adalah record yang jika dihapus akan **mengubah persamaan regresi secara signifikan**. Record ini tidak selalu punya residual besar. Ukuran yang dipakai:
- **Leverage (hat value)**: seberapa jauh nilai X sebuah record dari rata-rata X. Nilai di atas $2(P+1)/n$ dianggap tinggi.
- **Cook's distance**: menggabungkan leverage dan residual. Nilai di atas $4/(n-P-1)$ dianggap berpengaruh.

Contoh sintetis berikut menunjukkan **satu titik ekstrem** (8, 8) yang mengubah garis regresi secara drastis. Garis solid memakai semua data, sedangkan garis putus-putus tanpa titik tersebut.

```python
%matplotlib inline
from scipy.stats import linregress

np.random.seed(5)
x = np.random.normal(size=25)
y = -x / 5 + np.random.normal(size=25)
x[0] = 8
y[0] = 8

def abline(slope, intercept, ax):
    """Calculate coordinates of a line based on slope and intercept"""
    x_vals = np.array(ax.get_xlim())
    return (x_vals, intercept + slope * x_vals)

fig, ax = plt.subplots(figsize=(4, 4))
ax.scatter(x, y)
slope, intercept, _, _, _ = linregress(x, y)
ax.plot(*abline(slope, intercept, ax))
slope, intercept, _, _, _ = linregress(x[1:], y[1:])
ax.plot(*abline(slope, intercept, ax), '--')
ax.set_xlim(-2.5, 8.5)
ax.set_ylim(-2.5, 8.5)

plt.tight_layout()
plt.show()
```

**Interpretasi:** Satu titik di (8, 8) **menarik garis regresi** (garis solid) secara drastis, sehingga kemiringannya menjadi positif. Tanpa titik tersebut, garisnya (putus-putus) hampir datar atau sedikit negatif. Titik ini memiliki **leverage tinggi** karena nilai X-nya jauh dari data lain.

**Bubble plot** influence: sumbu-x adalah hat value (leverage), sumbu-y adalah standardized residual, dan ukuran lingkaran sebanding dengan Cook's distance.

```python
influence = OLSInfluence(result_98105)
fig, ax = plt.subplots(figsize=(5, 5))
ax.axhline(-2.5, linestyle='--', color='C1')
ax.axhline(2.5, linestyle='--', color='C1')
ax.scatter(influence.hat_matrix_diag, influence.resid_studentized_internal, 
           s=1000 * np.sqrt(influence.cooks_distance[0]),
           alpha=0.5)

ax.set_xlabel('hat values')
ax.set_ylabel('studentized residuals')

plt.tight_layout()
plt.show()
```

**Interpretasi:** Sebagian besar rumah memiliki leverage kecil dan residual dalam batas ±2,5 (garis putus-putus). Beberapa titik memiliki **lingkaran besar** (Cook's distance tinggi), yaitu record yang paling memengaruhi model.

```python
mask = [dist < .08 for dist in influence.cooks_distance[0]]
house_infl = house_98105.loc[mask]

ols_infl = sm.OLS(house_infl[outcome], house_infl[predictors].assign(const=1))
result_infl = ols_infl.fit()

pd.DataFrame({
    'Original': result_98105.params,
    'Influential removed': result_infl.params,
})
```

**Interpretasi:** Setelah record berpengaruh (Cook's distance ≥ 0,08) dihapus, koefisien berubah cukup besar. `Bathrooms` berubah dari **+2.282 menjadi −16.132** (berganti tanda), dan `SqFtTotLiving` naik dari 209,6 menjadi 230,1. Ini menunjukkan bahwa sedikit record saja bisa sangat memengaruhi interpretasi model, terutama pada data kecil (313 rumah). Pada data besar, efek seperti ini biasanya lebih kecil.

### Heteroskedasticity, Non-Normality, and Correlated Errors

**Teori:**
- **Heteroskedasticity**: variance residual tidak konstan. Misalnya, error semakin besar untuk rumah yang lebih mahal. Akibatnya, p-value dan CI tidak akurat.
- **Non-normality**: distribusi residual tidak normal (misalnya long-tailed).
- **Correlated errors**: error saling berkorelasi, misalnya pada data *time series*. Kondisi ini bisa dideteksi dengan statistik **Durbin-Watson**.

`regplot` dari *seaborn* menambahkan garis **lowess** (*smoother*) pada plot |residual| vs fitted value untuk melihat tren variance:

```python
fig, ax = plt.subplots(figsize=(5, 5))
sns.regplot(x=result_98105.fittedvalues, y=np.abs(result_98105.resid), 
            scatter_kws={'alpha': 0.25},
            line_kws={'color': 'C1'},
            lowess=True, ax=ax)
ax.set_xlabel('predicted')
ax.set_ylabel('abs(residual)')

plt.tight_layout()
plt.show()
```

**Interpretasi:** Garis lowess **naik ke kanan**: semakin tinggi harga prediksi, semakin besar absolute residual-nya. Ini adalah tanda **heteroskedasticity**. Model kurang akurat untuk rumah mahal, mungkin karena ada faktor yang belum tertangkap untuk segmen tersebut.

```python
fig, ax = plt.subplots(figsize=(4, 4))
pd.Series(influence.resid_studentized_internal).hist(ax=ax)
ax.set_xlabel('std. residual')
ax.set_ylabel('Frequency')


plt.tight_layout()
plt.show()
```

## 4.7 Polynomial and Spline Regression
**Teori:** Hubungan X dan Y sering **tidak linear**. Beberapa cara memodelkannya tanpa meninggalkan kerangka regresi:

### Polynomial Regression
Tambahkan suku pangkat dari prediktor:
$$Y = b_0 + b_1 X + b_2 X^2 + e$$
Model ini tetap disebut "linear" karena **linear terhadap koefisiennya**, meskipun kurvanya tidak lurus.

```python
model_poly = smf.ols(formula='AdjSalePrice ~  SqFtTotLiving + np.power(SqFtTotLiving, 2) + ' + 
                'SqFtLot + Bathrooms + Bedrooms + BldgGrade', data=house_98105)
result_poly = model_poly.fit()
print(result_poly.summary())
```

**Interpretasi:** Koefisien suku kuadrat `np.power(SqFtTotLiving, 2)` = **0,0388** dengan p-value ≈ 0,000 (**signifikan**). Artinya, hubungan luas rumah dengan harga **melengkung ke atas**: harga per sqft makin tinggi untuk rumah yang lebih besar. R² naik sedikit dari 0,795 menjadi **0,806**.

Implementasi partial residual plot untuk suku non-linear, karena `statsmodels` hanya mendukung suku linear:

```python
def partialResidualPlot(model, df, outcome, feature, ax):
    y_pred = model.predict(df)
    # determine columns required for model
    required = set(model.params.index).intersection(df.columns)
    required.add(feature)
    copy_df = df[list(required)].copy().astype('float')
    for c in copy_df.columns:
        if c == feature:
            continue
        copy_df.loc[:, c] = 0.0
    feature_prediction = model.predict(copy_df)
    results = pd.DataFrame({
        'feature': df[feature],
        'residual': df[outcome] - y_pred,
        'ypartial': feature_prediction - model.params.iloc[0],
    })
    results = results.sort_values(by=['feature'])
    smoothed = sm.nonparametric.lowess(results.ypartial + results.residual,
                                       results.feature, frac=1/3)
    
    ax.scatter(results.feature, results.ypartial + results.residual)
    ax.plot(smoothed[:, 0], smoothed[:, 1], color='gray')
    ax.plot(results.feature, results.ypartial, color='black')
    ax.set_xlabel(feature)
    ax.set_ylabel(f'Residual + {feature} contribution')
    return ax

fig, ax = plt.subplots(figsize=(5, 5))
partialResidualPlot(result_poly, house_98105, 'AdjSalePrice', 'SqFtTotLiving', ax)

plt.tight_layout()
plt.show()
print(result_poly.params.iloc[2])
```

**Interpretasi:** Partial residual plot model polynomial memperlihatkan kurva yang **melengkung ke atas** dan lebih mengikuti pola titik-titik dibanding garis lurus.

### Splines
**Teori:** Polynomial berderajat tinggi bisa "liar" di ujung-ujung data. **Spline** adalah solusinya: rentang X dibagi oleh beberapa **knots**, lalu setiap segmen dipasangi polynomial kecil (biasanya kubik) yang tersambung secara mulus di titik-titik knot. Di `statsmodels`, spline dibuat dengan fungsi `bs()` (*B-spline*), dengan `df` mengatur jumlah knot dan `degree=3` berarti kubik.

```python
formula = ('AdjSalePrice ~ bs(SqFtTotLiving, df=6, degree=3) + ' + 
           'SqFtLot + Bathrooms + Bedrooms + BldgGrade')
model_spline = smf.ols(formula=formula, data=house_98105)
result_spline = model_spline.fit()
print(result_spline.summary())
```

**Interpretasi:** Model spline (B-spline kubik dengan df = 6) menaikkan R² menjadi **0,814**, sedikit lebih tinggi dari polynomial.

```python
fig, ax = plt.subplots(figsize=(5, 5))
partialResidualPlot(result_spline, house_98105, 'AdjSalePrice', 'SqFtTotLiving', ax)

plt.tight_layout()
plt.show()
```

**Interpretasi:** Kurva spline lebih fleksibel. Kurva ini bisa datar di satu rentang lalu naik tajam di rentang lain. Namun, kita perlu berhati-hati agar tidak **overfitting**, terutama di ujung-ujung data yang jumlahnya sedikit.

### Generalized Additive Models (GAM)
**Teori:** Pada spline, kita harus menentukan knot sendiri. **GAM** mencari bentuk kurva secara **otomatis** untuk setiap prediktor:
$$Y = b_0 + f_1(X_1) + f_2(X_2) + \dots + e$$
Setiap $f_j$ adalah fungsi halus (*smooth*) yang diestimasi dari data. Model ini fleksibel tetapi tetap mudah diinterpretasikan karena efeknya bersifat aditif.

#### GAM dengan statsmodels

```python
from statsmodels.gam.api import GLMGam, BSplines

predictors = ['SqFtTotLiving', 'SqFtLot', 'Bathrooms', 'Bedrooms', 'BldgGrade']
outcome = 'AdjSalePrice'

x_spline = house_98105[predictors]
bs = BSplines(x_spline, df=[10] + [3] * 4, degree=[3] + [2] * 4)
# penalization weight
alpha = np.array([0] * 5)

formula = ('AdjSalePrice ~ SqFtTotLiving + ' + 
           'SqFtLot + Bathrooms + Bedrooms + BldgGrade')

gam_sm = GLMGam.from_formula(formula, data=house_98105, smoother=bs, alpha=alpha)
res_sm = gam_sm.fit()
print(res_sm.summary())
```

**Interpretasi:** Model GAM `statsmodels` memakai B-spline untuk `SqFtTotLiving` sambil memperlakukan prediktor lain secara linear.

```python
res_sm.plot_partial(0, cpr=True);
```

**Interpretasi:** Partial plot GAM untuk `SqFtTotLiving` menunjukkan kurva non-linear yang ditemukan **otomatis** dari data, dengan bentuk serupa model spline.

#### GAM dengan pyGAM

```python
predictors = ['SqFtTotLiving', 'SqFtLot', 'Bathrooms', 
              'Bedrooms', 'BldgGrade']
outcome = 'AdjSalePrice'
X = house_98105[predictors].values
y = house_98105[outcome]

## model
gam = LinearGAM(s(0, n_splines=12) + l(1) + l(2) + l(3) + l(4))
gam.gridsearch(X, y)
print(gam.summary())
```

```python
fig, axes = plt.subplots(figsize=(8, 8), ncols=2, nrows=3)

titles = ['SqFtTotLiving', 'SqFtLot', 'Bathrooms', 'Bedrooms', 'BldgGrade']
for i, title in enumerate(titles):
    ax = axes[i // 2, i % 2]
    XX = gam.generate_X_grid(term=i)
    ax.plot(XX[:, i], gam.partial_dependence(term=i, X=XX))
    ax.plot(XX[:, i], gam.partial_dependence(term=i, X=XX, width=.95)[1], c='r', ls='--')
    ax.set_title(titles[i]);
    
axes[2][1].set_visible(False)

plt.tight_layout()
plt.show()
```

**Interpretasi:** Pada model pyGAM, hanya `SqFtTotLiving` yang dimodelkan dengan fungsi *smooth* `s()`, sedangkan prediktor lain linear `l()`. Kurva `SqFtTotLiving` berbentuk **S (non-linear)**: landai pada rumah kecil, naik tajam pada rentang ±2.500–4.000 sqft, lalu sedikit melandai lagi. `BldgGrade` naik secara linear dan kuat. `Bathrooms` hampir datar, dan `Bedrooms` sedikit menurun. Garis merah putus-putus adalah confidence interval, yang semakin lebar di area dengan sedikit data.

## Materi Tambahan (tidak ada di buku): Regularization — Lasso
**Teori:** **Lasso** (*Least Absolute Shrinkage and Selection Operator*) menambahkan penalti $\alpha \sum |b_j|$ pada fungsi loss OLS. Efeknya, koefisien yang tidak penting **ditekan sampai tepat nol**, sehingga Lasso sekaligus melakukan **seleksi variabel**. Prediktor perlu di-*standardize* terlebih dahulu agar penalti berlaku adil, dan nilai $\alpha$ terbaik dipilih dengan cross-validation (`LassoCV`).

```python
from sklearn.linear_model import Lasso, LassoLars, LassoCV, LassoLarsCV
from sklearn.preprocessing import StandardScaler
```

```python
subset = ['AdjSalePrice', 'SqFtTotLiving', 'SqFtLot', 'Bathrooms', 
          'Bedrooms', 'BldgGrade']

house = pd.read_csv(HOUSE_CSV, sep='\t')
print(house[subset].head())
```

```python
predictors = ['SqFtTotLiving', 'SqFtLot', 'Bathrooms', 'Bedrooms',
              'BldgGrade', 'PropertyType', 'NbrLivingUnits',
              'SqFtFinBasement', 'YrBuilt', 'YrRenovated', 
              'NewConstruction']
outcome = 'AdjSalePrice'

X = pd.get_dummies(house[predictors], drop_first=True)
X['NewConstruction'] = [1 if nc else 0 for nc in X['NewConstruction']]
columns = X.columns
# X = StandardScaler().fit_transform(X * 1.0)
y = house[outcome]

house_lm = LinearRegression()
print(house_lm.fit(X, y))
```

```python
house_lasso = Lasso(alpha=10)
print(house_lasso.fit(X, y))
```

```python
Method = LassoLars
MethodCV = LassoLarsCV
Method = Lasso
MethodCV = LassoCV

alpha_values = []
results = []
for alpha in [0.001, 0.01, 0.1, 1, 10, 100, 1000, 10000, 100000, 1000000, 10000000]:
    model = Method(alpha=alpha)
    model.fit(X, y)
    alpha_values.append(alpha)
    results.append(model.coef_)
modelCV = MethodCV(cv=5)
modelCV.fit(X, y)
ax = pd.DataFrame(results, index=alpha_values, columns=columns).plot(logx=True, legend=False)
ax.axvline(modelCV.alpha_)
plt.show()
```

**Interpretasi:** Grafik *coefficient path* menunjukkan nilai setiap koefisien (garis berwarna) untuk berbagai nilai $\alpha$ (skala log). Untuk $\alpha$ kecil, koefisien sama seperti OLS biasa. Semakin besar $\alpha$, koefisien semakin **ditekan menuju nol**, sampai akhirnya hampir semuanya bernilai 0. Garis vertikal biru menandai $\alpha$ terbaik hasil cross-validation (`LassoCV`).

```python
pd.DataFrame({
    'name': columns,
    'coef': modelCV.coef_, 
})
```

**Interpretasi:** Pada $\alpha$ hasil cross-validation, hanya `SqFtTotLiving`, `SqFtLot`, `SqFtFinBasement`, dan `YrRenovated` yang koefisiennya tidak nol. Variabel lain, termasuk `Bathrooms`, `Bedrooms`, dan bahkan `BldgGrade`, ditekan menjadi **tepat 0** sehingga otomatis tereliminasi. Ini menunjukkan Lasso sebagai alat **seleksi variabel**. Karena di sini prediktor tidak distandardisasi, variabel dengan skala besar (seperti luas dalam sqft) cenderung lebih "tahan" terhadap penalti. Inilah alasan standardisasi direkomendasikan sebelum memakai Lasso.

## Ringkasan Chapter 4
1. **Simple linear regression** ($Y = b_0 + b_1X + e$) mengkuantifikasi hubungan satu prediktor dengan outcome. Koefisien diestimasi dengan **least squares**, yaitu meminimalkan jumlah kuadrat residual.
2. **Multiple linear regression** memakai banyak prediktor. Setiap koefisien diinterpretasikan **dengan prediktor lain tetap**.
3. **Evaluasi model:** **RMSE/RSE** untuk akurasi prediksi, **R²** untuk proporsi variasi yang dijelaskan, serta **t-stat/p-value** untuk signifikansi koefisien.
4. **Model selection:** Adjusted R², **AIC**, dan BIC memberi penalti pada kompleksitas. **Stepwise regression** dan regularization (**Lasso**) membantu memilih variabel dan mencegah overfitting.
5. **Weighted regression** memberi bobot berbeda pada setiap record.
6. **Prediction interval** (untuk nilai individual) selalu lebih lebar dari **confidence interval** (untuk rata-rata). Hindari **ekstrapolasi** di luar rentang data.
7. **Variabel kategorikal** diubah menjadi **dummy** dengan $k-1$ kategori (reference coding). Kategori yang terlalu banyak bisa **dikelompokkan** (contoh: ZipGroup).
8. **Interpretasi koefisien** dipengaruhi oleh **korelasi antar prediktor** (multicollinearity), **confounding variable**, dan **interaksi**.
9. **Diagnostics:** cek **outlier** (standardized residual), **influential values** (leverage, Cook's distance), **heteroskedasticity**, non-normality, dan **partial residual plot** untuk mendeteksi non-linearitas.
10. **Hubungan non-linear** dapat dimodelkan dengan **polynomial**, **spline**, dan **GAM**, sambil tetap berada dalam kerangka regresi.

**Kaitan dengan Machine Learning:** regresi linear adalah *baseline model* untuk masalah prediksi numerik. Konsep RMSE, overfitting, cross-validation, regularization, dan feature engineering (dummy, interaksi, polynomial) dipakai di hampir semua algoritma ML.
